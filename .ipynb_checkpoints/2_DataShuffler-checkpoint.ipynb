# Making the Training Data Set 

After the numerical simulation, the data should be shuffled to construct the training data set. 

```python
import os
commandor=''
for i in range(0,40):
    commandor=commandor+'python TDHiShuffler.py '+str(i)+'\n'
with open('Shuffler1.in','w') as writter:
    writter.write(commandor)
```

```python
import os
commandor=''
for i in range(0,100):
    commandor=commandor+'python TDHiShuffler2.py '+str(i)+'\n'
with open('Shuffler2.in','w') as writter:
    writter.write(commandor)
```
