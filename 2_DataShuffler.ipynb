# Making the Training Data Set 

After the numerical simulation, the data should be shuffled to construct the training data set. 

In the paper, there are 320 numerical simulation instances calculated in four batches via the TAMU HPRC job submission system. The numerical simulation results are stored in the "FiveHWRunner" directory, with the file names in the format "storeField_x_y.h5", where "x" is the batch number (from 0 to 3), "y" is the index number (from 0 to 79). The first three batches are used for the training data set, the fourth batch is for the testing data set. 

The numerical simulation results require about 2TB of the disk space. The training data set shuffling is done in two steps due to the data size. In the first step with "TDHiShuffler.py" script, only the plasma state in the turbulent phase are selected, which is approximately after t=100. Then, the initial plasma state is randomly selected in the plasma state time sequence. The terminal plasma state is also randomly selected between 0 and 200 time steps (between t=0 and t=1). All the plasma snapshots are only used ones. When all the plasma snapshots are used, or the remaining plasma snapshots cannot construct plasma state pairs, the first step of shuffling is finished, the shuffled plasma state pairs are stored in the directory "Cache/". 

In the second step with "TDHiShuffler2.py" of shuffling, The shuffled data pairs of different numerical simulation instances are further mixed, and stored into 100 separate files in the directory "TDHiData/". 

```python
import os
commandor=''
for i in range(0,40):
    commandor=commandor+'python TDHiShuffler.py '+str(i)+'\n'
with open('Shuffler1.in','w') as writter:
    writter.write(commandor)
```

```python
import os
commandor=''
for i in range(0,100):
    commandor=commandor+'python TDHiShuffler2.py '+str(i)+'\n'
with open('Shuffler2.in','w') as writter:
    writter.write(commandor)
```
